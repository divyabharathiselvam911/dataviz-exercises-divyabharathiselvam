## Task 1 - A line chart showing CO2 per Capita for all countries in the dataset

**Requirements:**
- All countries shown (for context), but only **one highlighted in colour** — your choice which
- All other lines in grey (#EEEEEE), thinner
- Highlighted country **labelled directly** at the end of its line (not in a legend)
- Insight title that names the highlighted country and its story

```python

import plotly.graph_objects as go

highlight_country = "China"

fig = go.Figure()

# All countries in light grey
for country in df["Country"].unique():
    if country == highlight_country:
        continue

    country_data = df[df["Country"] == country].sort_values("Year")

    fig.add_trace(go.Scatter(
        x=country_data["Year"],
        y=country_data["CO2_per_capita"],
        mode="lines",
        line=dict(color="#EEEEEE", width=1),
        showlegend=False,
        hoverinfo="skip"
    ))

# Highlight China
china = df[df["Country"] == highlight_country].sort_values("Year")

fig.add_trace(go.Scatter(
    x=china["Year"],
    y=china["CO2_per_capita"],
    mode="lines+text",
    line=dict(color="#E4572E", width=3),
    text=[""] * (len(china) - 1) + ["China"],
    textposition="middle right",
    showlegend=False
))

fig.update_layout(
    title="China's CO₂ per capita from 2000 to 2022",
    xaxis_title="Year",
    yaxis_title="CO₂ per capita",
    template="plotly_white",
    showlegend=False,
    margin=dict(r=90)
)

fig.update_xaxes(showgrid=False)
fig.update_yaxes(showgrid=True, gridcolor="#EEEEEE")

fig.show()
```

## Task 2 — A line chart showing CO2 emissions over time for all Asian countries in the dataset, with one country highlighted.

**same requirements as Task1**

```python


import plotly.graph_objects as go

# Select Asian countries only
asia = df[df["Region"] == "Asia"]

highlight_country = "China"

fig = go.Figure()

# All Asian countries in light grey
for country in asia["Country"].unique():
    if country == highlight_country:
        continue

    country_data = asia[asia["Country"] == country].sort_values("Year")

    fig.add_trace(go.Scatter(
        x=country_data["Year"],
        y=country_data["CO2_Mt"],
        mode="lines",
        line=dict(color="#EEEEEE", width=1),
        showlegend=False,
        hoverinfo="skip"
    ))

# Highlight China
china = asia[asia["Country"] == highlight_country].sort_values("Year")

fig.add_trace(go.Scatter(
    x=china["Year"],
    y=china["CO2_Mt"],
    mode="lines+text",
    line=dict(color="#E4572E", width=3),
    text=[""] * (len(china) - 1) + ["China"],
    textposition="middle right",
    showlegend=False
))

fig.update_layout(
    title="China's CO₂ emissions rose sharply from 2000 to 2022",
    xaxis_title="Year",
    yaxis_title="CO₂ emissions (Mt)",
    template="plotly_white",
    showlegend=False,
    margin=dict(r=90)
)

fig.update_xaxes(showgrid=False)
fig.update_yaxes(showgrid=True, gridcolor="#EEEEEE")

fig.show()
```
